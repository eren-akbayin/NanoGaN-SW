```python
import numpy as np
import plotly.graph_objects as go
```

```python
time = np.arange(0, 2000) * 1e-3

data = np.fromfile('measurements/angle_meas13.bin', dtype=np.float32)

if len(data) >= 6000:
    angles = data[:6000].reshape(3, 2000)
    angle1, angle2, angle3 = angles[0], angles[1], angles[2]
else:
    angle1 = data[0:2000]
    angle2 = data[2000:4000]
    angle3 = data[4000:6000]

print(f"Loaded {len(data)} samples")
print(f"angle1: {angle1.shape}, angle2: {angle2.shape}, angle3: {angle3.shape}")
```

```python
fig = go.Figure()

for angle, name in zip([angle1, angle2, angle3], ['Angle 1', 'Angle 2', 'Angle 3']):
    fig.add_trace(go.Scatter(
        x=time, y=angle,
        mode='lines',
        name=name,
        hovertemplate=f'Time: %{{x:.3f}}s<br>{name}: %{{y:.4f}}<extra></extra>'
    ))

fig.update_layout(
    title='Angle Measurements vs Time',
    xaxis_title='Time (seconds)',
    yaxis_title='Angle',
    hovermode='x unified',
    legend=dict(x=1, xanchor='right'),
    template='plotly_white'
)

fig.show()
```
